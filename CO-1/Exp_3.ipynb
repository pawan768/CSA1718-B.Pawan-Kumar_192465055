```python
from collections import deque
CAP_A, CAP_B = 4, 3
start = (0, 0)
queue = deque([(start, [])])
visited = {start}

while queue:
    (a, b), path = queue.popleft()
    if a == 2:
        print("Steps (amount in 4L jug, amount in 3L jug):")
        for state in path + [(a, b)]:
            print(state)
        break

    next_states = [
        (CAP_A, b),
        (a, CAP_B),
        (0, b),
        (a, 0),
        (a - min(a, CAP_B - b), b + min(a, CAP_B - b)),
        (a + min(b, CAP_A - a), b - min(b, CAP_A - a))
    ]

    for state in next_states:
        if state not in visited:
            visited.add(state)
            queue.append((state, path + [(a, b)]))
else:
    print("No solution found.")
```
