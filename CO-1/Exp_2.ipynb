```python
N = 8
board = [-1] * N

def solve(row):
    if row == N:
        print("One solution:")
        for col in board:
            print("." * col + "Q" + "." * (N - col - 1))
        return True

    for col in range(N):
        safe = True
        for previous_row in range(row):
            previous_col = board[previous_row]
            if (previous_col == col or
                    abs(previous_col - col) == abs(previous_row - row)):
                safe = False
                break

        if safe:
            board[row] = col
            if solve(row + 1):
                return True
    return False

solve(0)
```
