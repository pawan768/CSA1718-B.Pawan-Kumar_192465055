```python
from itertools import permutations

cost = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(cost)
best_cost = float("inf")
best_path = None

# Keep city 0 as the starting and ending city.
for order in permutations(range(1, n)):
    path = (0,) + order + (0,)
    total = sum(cost[path[i]][path[i + 1]] for i in range(n))

    if total < best_cost:
        best_cost = total
        best_path = path

print("Best route (city numbers):", " -> ".join(map(str, best_path)))
print("Minimum travel cost:", best_cost)
```
