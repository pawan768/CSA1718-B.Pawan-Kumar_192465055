```python
letters = "SENDMORY"
assignment = {}

def solve():
    def backtrack(index, used):
        if index == len(letters):
            d = assignment
            send = 1000*d["S"] + 100*d["E"] + 10*d["N"] + d["D"]
            more = 1000*d["M"] + 100*d["O"] + 10*d["R"] + d["E"]
            money = 10000*d["M"] + 1000*d["O"] + 100*d["N"] + 10*d["E"] + d["Y"]
            return send + more == money

        letter = letters[index]
        for digit in range(10):
            if digit in used:
                continue
            if letter in ("S", "M") and digit == 0:
                continue
            assignment[letter] = digit
            if backtrack(index + 1, used | {digit}):
                return True
            del assignment[letter]
        return False

    if backtrack(0, set()):
        d = assignment
        send = 1000*d["S"] + 100*d["E"] + 10*d["N"] + d["D"]
        more = 1000*d["M"] + 100*d["O"] + 10*d["R"] + d["E"]
        money = 10000*d["M"] + 1000*d["O"] + 100*d["N"] + 10*d["E"] + d["Y"]
        print(f"{send} + {more} = {money}")
        print("Digit assignment:", d)
    else:
        print("No solution found.")

solve()
```
