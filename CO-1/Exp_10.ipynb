```python
import heapq

graph = {
    "A": [("B", 1), ("C", 4)],
    "B": [("C", 2), ("D", 5)],
    "C": [("D", 1)],
    "D": []
}

# Estimated distance from each node to goal D
heuristic = {"A": 4, "B": 3, "C": 1, "D": 0}

def astar(start, goal):
    # Priority queue entries: (f = g + h, g, node, path)
    queue = [(heuristic[start], 0, start, [start])]
    best_g = {}

    while queue:
        f, g, node, path = heapq.heappop(queue)

        if node == goal:
            print("Shortest path:", " -> ".join(path))
            print("Total cost:", g)
            return

        if node in best_g and best_g[node] <= g:
            continue
        best_g[node] = g

        for neighbour, edge_cost in graph[node]:
            new_g = g + edge_cost
            if neighbour not in best_g or new_g < best_g[neighbour]:
                heapq.heappush(
                    queue,
                    (new_g + heuristic[neighbour], new_g,
                     neighbour, path + [neighbour])
                )

    print("No path found.")

astar("A", "D")
```
