```python
from collections import deque

# State: (missionaries on left, cannibals on left, boat side)
start = (3, 3, 1)  # 1 = left, 0 = right
goal = (0, 0, 0)
boat_moves = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]

def valid(m, c):
    if not (0 <= m <= 3 and 0 <= c <= 3):
        return False
    right_m, right_c = 3 - m, 3 - c
    return (m == 0 or m >= c) and (right_m == 0 or right_m >= right_c)

queue = deque([(start, [])])
visited = {start}

while queue:
    state, path = queue.popleft()
    if state == goal:
        print("State: (missionaries left, cannibals left, boat side)")
        for s in path + [state]:
            print(s)
        break

    m, c, boat = state
    for dm, dc in boat_moves:
        if boat == 1:
            nxt = (m - dm, c - dc, 0)
        else:
            nxt = (m + dm, c + dc, 1)

        if valid(nxt[0], nxt[1]) and nxt not in visited:
            visited.add(nxt)
            queue.append((nxt, path + [state]))
else:
    print("No solution found.")
```
